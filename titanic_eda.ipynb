## Task 02: Data Cleaning and Exploratory Data Analysis (EDA) on Titanic Dataset

**Objective**: Perform data cleaning and exploratory data analysis (EDA) on the Titanic dataset from Kaggle. Explore relationships between variables and identify patterns and trends that contributed to passenger survival.

### 1. Import Libraries and Load Data

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Visual formatting
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (8, 5)

# Load Titanic dataset
df = pd.read_csv("train.csv")
print(f"Dataset Dimensions: {df.shape[0]} rows, {df.shape[1]} columns")
df.head()
```

### 2. Dataset Overview and Missing Values Inspection

```
# Check structure and column types
df.info()
```

```
# Check missing values
missing = df.isnull().sum()
missing[missing > 0]
```

### 3. Data Cleaning and Feature Engineering
- **Age**: Impute missing values with the median age of respective passenger class (`Pclass`).
- **Embarked**: Impute missing values with the mode (`S`).
- **Cabin**: Missing in ~77% rows; drop column and create `Has_Cabin` flag.
- **FamilySize**: Derive from `SibSp + Parch + 1`.

```
# Impute missing Age
df['Age'] = df.groupby('Pclass')['Age'].transform(lambda x: x.fillna(x.median()))

# Impute missing Embarked
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Process Cabin
df['Has_Cabin'] = df['Cabin'].apply(lambda x: 0 if pd.isna(x) else 1)
df.drop(columns=['Cabin'], inplace=True)

# Feature Engineering
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1

# Confirm no missing values left
print("Missing values after cleaning:")
print(df.isnull().sum())
```

### 4. Exploratory Data Analysis (EDA) & Visualizations

#### 4.1 Survival Rate by Gender

```
plt.figure(figsize=(6, 4))
sns.barplot(data=df, x='Sex', y='Survived', hue='Sex', palette='Blues_d', errorbar=None, legend=False)
plt.title('Survival Rate by Gender', fontsize=12, fontweight='bold')
plt.ylabel('Survival Rate')
plt.xlabel('Gender')
plt.show()

# Survival percentage
print(df.groupby('Sex')['Survived'].mean().round(4) * 100)
```

#### 4.2 Survival Rate by Ticket Class (Pclass)

```
plt.figure(figsize=(6, 4))
sns.barplot(data=df, x='Pclass', y='Survived', hue='Pclass', palette='viridis', errorbar=None, legend=False)
plt.title('Survival Rate by Passenger Class', fontsize=12, fontweight='bold')
plt.ylabel('Survival Rate')
plt.xlabel('Passenger Class')
plt.show()

print(df.groupby('Pclass')['Survived'].mean().round(4) * 100)
```

#### 4.3 Age Distribution of Survivors vs Non-Survivors

```
plt.figure(figsize=(8, 4))
sns.histplot(data=df, x='Age', hue='Survived', kde=True, bins=30, palette='coolwarm', element='step')
plt.title('Age Distribution: Non-Survivors (0) vs Survivors (1)', fontsize=12, fontweight='bold')
plt.xlabel('Age')
plt.show()
```

#### 4.4 Correlation Heatmap

```
plt.figure(figsize=(8, 6))
numeric_cols = df.select_dtypes(include=[np.number]).columns
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='mako', fmt='.2f', linewidths=0.5)
plt.title('Correlation Matrix of Numerical Features', fontsize=12, fontweight='bold')
plt.show()
```

### 5. Key Findings & Insights
1. **Gender**: Female passengers had a ~74.2% survival rate compared to only ~18.9% for males.
2. **Class Privilege**: 1st class passengers survived at 63.0%, while 3rd class had only 24.2% survival.
3. **Age Priority**: Children under 10 had significantly higher survival odds due to the "women and children first" evacuation rule.
4. **Family Size**: Small families (2-4 members) survived at higher rates than passengers traveling alone or in large families.
